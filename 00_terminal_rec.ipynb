# part1
## 核心操作命令流水線
### 1. 檢視與切換專案
git clone https://github.com/openJiuwen-ai/CareerSim-BDCI26.git
cd CareerSim-BDCI26
tree
git status

### 2. 環境同步與驗證
make sync
uv --version
uv run python --version
uv run python -m career_sim_runner --help

### 3. 測試 Ollama 服務與模型
ollama list
ollama serve
ollama run qwen3.5:9b
### (測試 prompt: "Reply only with OK")

curl http://localhost:11434/v1/chat/completions \
  -H "Content-Type: application/json" \
  -d '{
    "model": "qwen3.5:9b",
    "messages": [
      {
        "role": "user",
        "content": "Reply only with OK."
      }
    ]
  }'

### 4. 配置環境變數
cp .env.example .env
cat .env
# 配置為:
# API_BASE=http://localhost:11434/v1
# API_KEY=ollama
# MODEL_NAME=qwen3.5:9b

### 5. 運行 Setup 與啟動 JiuwenSwarm
make setup
make start-jiuwen
uv run jiuwenswarm-start --list

### 6. 建構 solution 目錄結構與檔案
mkdir -p solution/design
mkdir -p solution/skills/career-survival-strategy
nano solution/manifest.json
nano solution/skills/career-survival-strategy/SKILL.md
nano solution/README.md
nano solution/design/skills_design.md
nano solution/design/decision_report.md

### 7. 驗證提交格式
git status
git diff solution/manifest.json
uv run python -m career_sim_runner validate --submission solution

## 待修復項提示
solution/skills/career-survival-strategy/SKILL.md 缺少 YAML Frontmatter：
檔案頂部必須以 --- 開頭並包含元資料，例如：

```
---
name: career-survival-strategy
description: Career survival and long-horizon decision strategy.
---
# Career Survival Strategy
...
```

缺少數據集：
執行以下命令更新數據集：
```
uv run career-emulator update
```
