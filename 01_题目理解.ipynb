# 游戏本身
## 基本规则
Agent 从 L1 新人开始，在公司里生活 48个月 = 4年
模拟器会不断给 Agent 当前状态和职场事件。Agent 必须自己决定下一步。

需要同时管理：
职级、财富、健康、尊严、技能、人脉、产出、风险等状态。
本质上是：
有限资源 + 隐藏机制 + 随机事件下的长期序贯决策问题。

## 每季规则
每到季度末，还有额外决策。
每季度有 3 Energy + 1 个季度主行动
主动规划：
Q1
│
├─ Energy 1
├─ Energy 2
├─ Energy 3
│
└─ 主行动 × 1

> 例如
> Energy
> ├─ 恢复健康
> ├─ 恢复健康
> └─ 建立人脉
> Main Action
> └─ 社交 / 学习 / 休息……


## 每半年
每 6 个月进行一次绩效评估，并结算绩效奖金
所以 Agent 应该有时间意识。
Month 1：
可以接受短期投入，换长期 Skill。

Month 5：
马上绩效了，可能需要补 Output。

Month 23：
如果某个指标已经危险，应该优先避免爆雷。

Month 47：
只剩一个月，一些需要 12 个月才能回报的投资价值就明显下降。

这就是典型的 long-horizon planning。

# API / LLM 在里面是什么角色？
可以从架构上理解。
你的：
API_BASE
API_KEY
MODEL_NAME


官方示例 .env 也是
```
API_BASE=""
API_KEY=""
MODEL_NAME="deepseek-v4-flash"
``` :chatgpt-content-reference{index="9"}


LLM负责：
> 理解信息 + 推理 + 调 Skill + 做决定。

Skill负责：
> **告诉 LLM 应该怎样推理。**

模拟器负责：
> 告诉 Agent 决策造成什么后果。

所以：

```text
LLM = 大脑基础能力
Skill = 你教给它的职业策略
Career Emulator = 世界
JiuwenSwarm = Agent 身体 / 运行框架
```
