# 单人 Agent 工作流
## 工作流
CareerSim 当前状态
        ↓
读取 observe()
        ↓
状态解析
        ↓
判断当前模式
        ↓
生成候选行动
        ↓
风险检查
        ↓
长期价值评估
        ↓
人脉 / 技能 / 晋升收益评估
        ↓
选择行动
        ↓
take_action()
        ↓
记录结果
        ↓
更新经验
        ↓
进入下一回合

## 策略逻辑
Layer 1  生存约束
Health 是否危险？

Layer 2  当前模式
恢复 / 正常 / 成长 / 晋升冲刺 / 探索

Layer 3  当前瓶颈
Health / Skill / Network / Output / Wealth

Layer 4  候选行动比较
收益、成本、风险、长期价值

Layer 5  决策
选择最符合当前阶段的行动

Layer 6  复盘
记录这次选择实际发生什么

## 分工
1. Risk Agent
>这个选择会不会把人搞崩？
重点关注：
- Health
- 连续高压
- Dignity
- 不可逆损失
- 失败最坏情况


2. Career Agent
>这个选择是否有利于未来发展？
关注：
- Skill
- Output
- 晋升
- 绩效
- 当前职级
- 晋升窗口

3. Network Agent
>这个选择有没有建立有效人脉的价值？
关注：
- 合作关系
- 关键人物
- 信息价值
- 长期资源
- 是否只是无效社交

4. Planner / Leader
综合：
Risk Agent:
A风险低
B风险高
C风险中

Career Agent:
A职业价值4
B职业价值9
C职业价值6

Network Agent:
A人脉价值8
B人脉价值2
C人脉价值7

根据你的总策略：
Health > Network ≈ Skill > 晋升 > Wealth
