# CareerSim 状态机 v1

## 一、核心状态优先级

总体优先级：

**健康 > 晋升 > 财富**

核心长期资产：

**健康、人脉、技能**

默认行为风格：

**健康优先、提前规划、允许探索、避免崩盘。**

---

# 二、Health 状态机

## 1. 安全区：Health > 60

可以正常执行：

- 工作
- 学习
- 社交
- 探索未知事件
- 承担适度风险

但高压行为不应连续堆叠。

原则：

> 高压行为之后，应优先安排恢复，而不是连续透支。

即使 Health 很高，也不鼓励连续多轮压榨。

---

## 2. 警戒区：40 < Health ≤ 60

进入健康警戒模式。

此时：

- 降低高压任务优先级
- 不主动接受连续消耗
- 开始安排恢复
- 只有明显高价值机会才值得承担额外压力
- Skill / Network 增长仍可进行，但必须控制成本

原则：

> 此阶段不是完全停止发展，而是避免继续恶化。

---

## 3. 恢复区：Health ≤ 40

进入强制恢复模式。

此时：

- Health 成为绝对第一优先级
- 原则上停止高压工作
- 停止高健康损耗探索
- 暂缓非必要晋升冲刺
- 优先恢复至安全水平

只有非常特殊、极低健康成本且高收益的机会才考虑。

---

# 三、关键机会判断规则

对于一次性晋升机会或高收益机会，不采用“只要还能撑就上”的逻辑。

核心判断标准：

> 如果完成该行为后，预计需要超过一天才能恢复到正常工作状态，则直接放弃。

也就是说，判断重点不只是：

**会掉多少 Health**

而是：

**损耗后的恢复成本。**

这意味着 Agent 应考虑：

- 预计 Health 损失
- 恢复所需时间
- 恢复期间损失的其他机会
- 是否会影响后续连续行动

因此：

> 恢复成本本身就是风险成本。

---

# 四、绩效策略

不采用“DDL 冲刺型”策略。

只要知道未来存在绩效考核，就应开始逐步准备。

核心原则：

**提前积累，滚动调整。**

流程：

```text
已知未来有绩效
↓
提前提高 Output 关注度
↓
持续观察状态
↓
根据 Health / Skill / Network 调整
↓
临近考核只做微调
```

而不是：

```text
前面不管
↓
考核前疯狂冲刺
```

目标是保留：

- 调整空间
- 健康缓冲
- 意外事件应对能力
- 资源重新分配能力

---

# 五、Skill 与 Network 决策

当 Skill 和 Network 同时不足，而只能投资一个时：

**优先 Network。**

原因不是否定 Skill 的价值，而是你的策略更重视：

- 有效合作
- 长期机会
- 信息获取
- 晋升资源
- 关系网络

因此基本优先级：

**Network > Skill**

但如果 Skill 已成为明显的硬性晋升门槛，则可以动态调整。

---

# 六、短板策略

是否补短板，要结合剩余时间判断。

如果游戏早期：

- 长期回报明显
- 有足够时间回收投入

则可以补。

如果游戏后期：

- 补齐需要大量时间
- 回报无法在剩余月份兑现

则应放弃补齐。

因此短板判断公式应包含：

**收益 × 剩余可兑现时间**

而不是单纯看：

**现在低不低。**

---

# 七、财务保护模式

财富不使用固定金额作为警戒线，而采用历史动态标准。

进入财务保护模式的条件：

同时满足：

1. 当前存款 < 历史最高存款的 50%
2. 当前收入与支出状态已经无法维持每月储蓄 10% 月薪

此时进入：

**Financial Protection Mode**

主要行为：

- 降低非必要消费
- 降低高成本投资
- 暂缓收益不明确的支出
- 优先恢复现金流
- 重新建立储蓄能力

退出条件：

- 恢复稳定现金流
- 再次能够维持月薪 10% 储蓄

---

# 八、探索策略

即使进入最后 6 个月，也不强制降低探索欲望。

如果出现高潜力机会，仍然可以积极探索。

因此后期不是：

**全面求稳**

而是：

**更加严格筛选风险收益比。**

高收益、可承受失败的机会：

可以探索。

低价值、纯随机、失败代价大的机会：

减少探索。

---

# 九、财富换成长规则

如果一个选择能够同时提升：

- Skill
- Network

但会消耗财富，则只要财富仍处于安全状态，就愿意接受。

因此财富的作用是：

> 支撑成长，而不是单独追求最大化。

---

# 十、失败学习机制

失败后不直接给某个行动永久贴上：

**危险 / 禁止**

而是重新分析：

- 当时 Health 是否偏低
- 是否处于错误阶段
- 是否连续高压
- 是否缺乏 Network
- 是否 Skill 不足
- 是否只是随机事件
- 是否存在隐藏条件

核心学习单位：

**条件 + 行动 + 结果**

而不是：

**行动 = 成功 / 失败**

所以同一个事件，在不同状态下可以做不同选择。

---

# 十一、稳定状态下的默认行为

当同时满足：

- Health 良好
- Skill 良好
- Network 良好
- 当前没有明确晋升窗口

行为优先级为：

## 第一优先：D — 保持状态，减少无意义消耗

不为了“看起来很努力”而消耗资源。

## 第二优先：E — 主动探索高潜力机会

如果发现：

- 新项目
- 新人脉
- 新路径
- 新资源
- 高潜力事件

则可以主动尝试。

## 第三优先：B — 扩展有效人脉

没有明确高潜力机会时，继续增加高价值关系。

因此默认稳定状态逻辑是：

**保持 > 高潜力探索 > 有效人脉扩展**

---

# 十二、最终模式划分

整个 Agent 可以划分为 7 种主要模式：

### 1. Normal Mode
Health > 60，状态稳定。

执行正常工作、成长与探索。

### 2. Growth Mode
健康安全，当前存在明显 Skill / Network 成长机会。

优先长期资产。

### 3. Performance Preparation Mode
已知未来存在绩效考核。

提前积累 Output，不临时抱佛脚。

### 4. Health Warning Mode
40 < Health ≤ 60。

降低风险与连续高压行为。

### 5. Recovery Mode
Health ≤ 40。

强制优先恢复。

### 6. Financial Protection Mode
存款低于历史最高值 50%，并且无法持续储蓄月薪 10%。

恢复现金流。

### 7. Opportunity Exploration Mode
基础状态稳定，出现高潜力未知机会。

主动探索，但要求失败可承受。

---

# 十三、模式优先级

如果多个模式同时触发，则按照：

**Recovery**
>
**Health Warning**
>
**Financial Protection**
>
**Performance Preparation**
>
**Opportunity Exploration**
>
**Growth**
>
**Normal**

其中：

健康始终覆盖其他模式。

也就是说：

即使绩效考核临近，只要进入 Recovery Mode，就不允许被绩效目标覆盖。

---

# 十四、核心执行原则

最终每次决策都应问：

1. 会不会破坏健康安全？
2. 恢复成本高不高？
3. 是否有长期人脉价值？
4. 是否能提高技能？
5. 是否有助于未来晋升？
6. 是否影响滚动绩效准备？
7. 财富是否仍处安全区？
8. 失败是否可承受？
9. 当前剩余时间是否足以兑现收益？

最终目标：

> 不靠临时爆发，不靠过度压榨，而是通过健康缓冲、提前准备、有效人脉和持续技能积累，让 Agent 在 48 个月中保持长期行动能力。
