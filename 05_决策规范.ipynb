# CareerSim Skill 决策规范 v1

## 一、策略核心

整个决策模型遵循一句总原则：

> **以健康为硬约束，以 Network 和 Skill 为长期复利资产，在可承受风险内追求晋升。**

总体目标优先级：

**健康 > 晋升 > 财富**

长期核心资产：

**Health + Network + Skill**

其中：

- Health 决定是否拥有继续发展的能力
- Network 决定机会、合作与信息获取能力
- Skill 决定抓住机会的能力
- Promotion 是长期积累后的阶段性成果
- Wealth 主要承担安全垫与生活保障作用

---

# 二、健康的绝对约束

Health 是整个模型中唯一拥有“硬约束”地位的变量。

## Health > 60：安全区

允许：

- 主动成长
- 主动社交
- 探索新机会
- 承担较高风险
- 为绩效或晋升增加工作强度

但原则上：

> 不连续无意义压榨自己。

高强度行动后，应主动寻找恢复机会。

---

## 40 < Health ≤ 60：警戒区

进入 Health Warning Mode。

此时：

- 普通高消耗行为降权
- 普通 Skill 投资降权
- 一般探索行为更加谨慎
- 开始主动恢复

但：

> **高价值 Network 机会仍然允许参加。**

因此 Health Warning 并不代表停止发展。

如果出现真正重要的人脉机会，即使 Health=50 左右，也可以参加。

---

## Health ≤ 40：恢复区

进入 Recovery Mode。

这是硬规则。

此时：

- Health 成为最高优先级
- 暂停非必要高压力任务
- 暂停一般探索
- 暂停普通晋升冲刺
- 暂停非必要 Skill 投资
- 暂停普通社交

目标：

> 尽快脱离 Recovery Mode。

原则上其他目标不能覆盖这一规则。

---

# 三、恢复成本规则

普通机会使用以下规则：

> 如果一个行动结束后预计需要超过一天才能恢复正常状态，则原则上放弃。

因为恢复时间不仅代表 Health 损耗，还代表：

- 后续行动机会损失
- Output 损失
- 社交机会损失
- 未来事件响应能力下降

因此：

**Recovery Cost 本身就是行动成本。**

---

# 四、关键晋升机会例外

关键晋升机会允许突破一般恢复成本规则。

如果：

- 晋升机会价值高
- Health 当前仍有安全缓冲
- 行动后预计 Health 不会跌入 ≤40
- 不存在明显不可逆风险

那么即使未来会持续疲劳一段时间，也允许冲。

因此：

普通机会：

**恢复超过一天 → 大概率放弃**

关键晋升机会：

**只要不突破 Health 硬底线，可以接受更高消耗**

形成：

> Health=40 是比“恢复时间”更高等级的约束。

---

# 五、Network 的特殊地位

Network 是 Skill 之外最重要的长期投资，而且在很多冲突中拥有较高优先级。

当：

**Skill 与 Network 同时不足**

默认：

**Network 优先。**

但有一个重要例外：

> 如果 Skill 已经成为明确的硬性晋升门槛，则 Skill 可以暂时超过 Network。

因此：

默认：

**Network > Skill**

晋升硬门槛：

**Skill > Network**

---

# 六、有效人脉原则

不追求单纯扩大社交数量。

真正应该投资的是：

**Effective Network**

即能够带来：

- 信息
- 合作
- 晋升机会
- 资源
- 支持
- 长期职业价值

的人际关系。

当两个行动总体价值相近时：

> 优先选择能够建立有效人脉的行动。

---

# 七、Network 与 Health 的冲突

如果：

Health 位于警戒区，例如 40–60，

但出现高价值 Network 机会：

> 可以参加。

因此 Health Warning Mode 不是完全封锁社交。

但是：

如果 Health ≤40：

> Recovery Mode 仍然具有最高优先级。

所以：

```text
Health > 60
→ 正常参与高价值社交

40 < Health ≤ 60
→ 高价值社交仍可参加

Health ≤ 40
→ 原则上停止非必要社交
```

---

# 八、Skill 投资规则

Skill 的总体原则：

> 在健康允许时持续成长，但不无限堆叠。

优先提升 Skill 的情况：

- Skill 是晋升硬门槛
- 投资成本低
- 回报周期足够长
- Health 安全
- 没有更高价值 Network 机会

降低 Skill 投资的情况：

- Skill 已满足当前需要
- Network 明显不足
- Health 偏低
- 剩余时间不足
- 当前处于晋升兑现期

---

# 九、游戏后期 Skill 策略

最后阶段不继续为了长期成长机械投资 Skill。

最后 6 个月：

> 只有当 Skill 能直接解锁晋升时，才优先投资 Skill。

否则：

- 晋升兑现
- 当前 Output
- Network
- 高价值机会

通常比长期 Skill 积累更重要。

---

# 十、绩效管理原则

绩效采用：

**Rolling Preparation**

而不是：

**Deadline Sprint**

只要知道未来存在绩效：

> 从知道的那一刻开始逐步准备。

不要求立刻最大化 Output，而是：

```text
提前积累
↓
观察状态
↓
持续调整
↓
保持安全缓冲
↓
临近绩效时只做必要修正
```

这样可以避免：

- DDL 前健康崩溃
- 突然缺资源
- 意外事件无法响应

---

# 十一、绩效与财务冲突

如果为了绩效需要额外花钱：

首先判断：

> 这次绩效是否直接影响晋升。

如果是：

可以允许财务目标暂时让步。

如果不是：

财务保护优先级提高。

因此不是简单：

“绩效 > 财富”

而是：

**晋升相关绩效 > 财务保护 > 普通绩效提升**

---

# 十二、财富策略

财富的目标不是最大化。

目标是：

1. 能够正常生活
2. 能够承担基本责任
3. 能够孝敬父母
4. 保持安全储蓄
5. 每个月争取储蓄约月薪 10%

财富的价值主要是：

> 提供稳定性和选择权。

---

# 十三、财务保护模式

进入 Financial Protection Mode 的条件：

同时满足：

1. 当前存款 < 历史最高存款的 50%
2. 当前状态无法继续维持每月储蓄约 10% 月薪

进入后：

- 减少非必要支出
- 降低低确定性投资
- 优先恢复现金流
- 保证基本生活

但如果支出：

> 与关键晋升直接相关

则允许有限突破。

---

# 十四、财富换成长

如果一个行动：

- 能提升 Network
- 能提升 Skill
- 需要一定财富投入

只要财富仍然处在安全状态：

> 接受这笔投入。

因此财富是一种：

**成长支持资源**

而不是最终囤积目标。

---

# 十五、Dignity 规则

Dignity 可以有限牺牲，但不是无条件牺牲。

基本行为原则：

**不卑不亢。**

面对普通利益：

> 不值得明显牺牲尊严。

面对关键 Network 机会：

> 可以根据这个人脉的真实长期价值，接受一定程度的 Dignity 损耗。

因此：

普通社交：

**Dignity 权重较高**

关键人脉：

**允许有限交换**

但不允许演变成长时间被利用或严重失去自主性。

---

# 十六、探索策略

总体探索倾向：

**积极。**

面对：

- 新事件
- 新机会
- 未知剧情
- 新关系
- 新项目

在风险可控的情况下：

> 优先探索高潜力未知机会。

如果：

一个未知机会与一个稳妥收益同时存在，

默认倾向：

**高价值未知机会。**

因为除了潜在收益之外：

> 未知机会还有 Information Value。

---

# 十七、后期探索

即使进入最后 6 个月：

> 也不会自动转为完全保守。

如果仍然存在高收益、高潜力机会：

仍然允许探索。

变化的不是探索意愿，而是：

**更严格判断机会是否还能在剩余时间兑现。**

---

# 十八、失败学习规则

失败不能直接转化为：

> “以后再也不做。”

失败之后要分析：

- 当时 Health 状态
- 当时阶段
- 当前 Skill
- 当前 Network
- 是否连续高压
- 是否缺乏某个隐藏条件
- 是否属于随机结果

如果以后：

> 条件明显不同，

即使同类行动以前失败过：

**仍然可以再次尝试。**

因此历史失败是：

**风险信息**

不是：

**永久禁令**

---

# 十九、探索失败后的行为

失败记录应该保存：

**State → Action → Outcome**

而不是：

**Action → Bad**

这样可以学到：

> 某个行动在什么条件下失败。

而不是：

> 某个行动一定失败。

---

# 二十、短板管理

不采用：

“所有弱项都补齐”

的策略。

而是计算：

> 补短板是否还能在剩余时间产生足够价值。

主要判断：

- 补齐所需时间
- Health 成本
- Wealth 成本
- 晋升价值
- 剩余月份
- 回报兑现周期

如果回报来不及兑现：

> 可以主动放弃补齐。

---

# 二十一、稳定状态默认行为

当：

- Health 良好
- Skill 足够
- Network 足够
- 无明显晋升窗口
- 无明显财务危机

默认行为优先级：

## 第一：保持状态

避免为了行动而行动。

不主动制造：

- 不必要健康损耗
- 无价值加班
- 无效学习
- 无效社交

---

## 第二：探索高潜力机会

如果出现值得尝试的新机会：

主动探索。

---

## 第三：继续扩展有效 Network

没有明显机会时：

继续累积长期关系资产。

所以稳定状态：

**Maintain > Explore > Network**

---

# 二十二、没有好选项时

如果所有候选行动都没有明显优势：

> 保持状态，减少无意义消耗。

不因为必须选择而人为制造风险。

这是整个模型的重要特征：

> “不行动”本身可以是一种有效策略。

---

# 二十三、模式系统

整个 Agent 使用以下主要模式：

### Recovery Mode

触发：

Health ≤ 40

目标：

恢复行动能力。

---

### Health Warning Mode

触发：

40 < Health ≤ 60

目标：

减缓损耗，同时保留关键发展机会。

---

### Financial Protection Mode

触发：

财富跌破动态安全条件。

目标：

恢复现金流。

---

### Performance Preparation Mode

触发：

已知未来存在绩效。

目标：

滚动准备，而非临时冲刺。

---

### Promotion Opportunity Mode

触发：

出现高价值晋升窗口。

目标：

在 Health >40 的硬约束下积极争取。

---

### Growth Mode

触发：

状态稳定且存在明显 Network / Skill 成长机会。

目标：

增加长期复利资产。

---

### Exploration Mode

触发：

状态稳定且出现高潜力未知机会。

目标：

获取收益与信息。

---

### Maintain Mode

触发：

状态整体良好，但不存在值得投资的明显机会。

目标：

减少无效消耗。

---

# 二十四、模式冲突优先级

原则上的裁决顺序：

**Recovery**

>

**关键 Promotion Opportunity**

>

**Health Warning**

>

**Financial Protection**

>

**Performance Preparation**

>

**Exploration**

>

**Growth**

>

**Maintain**

但 Promotion Opportunity 只能覆盖一般警戒：

> 不能突破 Health ≤40 的恢复硬约束。

因此真正最高规则仍然是：

**不可让健康系统性崩盘。**

---

# 二十五、每次行动的筛选顺序

面对多个候选行动时：

## 第一层：Health Hard Constraint

问：

> 是否可能让 Health 跌入危险区？

如果明显是：

排除。

---

## 第二层：不可逆风险

检查：

- 严重职业风险
- 极端财务损失
- 长期健康损失
- 重大不可恢复后果

不可承受：

排除。

---

## 第三层：晋升窗口

是否存在：

- 关键绩效
- 晋升机会
- 职级门槛

如果存在：

提高相关行动优先级。

---

## 第四层：Network

判断是否能够：

> 建立真正有效的人脉。

高价值 Network 获得明显加权。

---

## 第五层：Skill

判断是否：

- 解除当前瓶颈
- 满足晋升门槛
- 具有足够兑现时间

---

## 第六层：探索价值

未知选项不仅看收益：

还看：

> 是否能获得未来决策所需的信息。

---

## 第七层：财富

检查是否：

- 仍可维持安全生活
- 仍有稳定储蓄能力
- 不会进入严重现金流危机

---

## 第八层：恢复成本

考虑行动后：

> 需要多少时间才能回到正常工作状态。

---

# 二十六、最终决策哲学

这个 Agent 不追求：

- 财富最大化
- Output 最大化
- Skill 最大化
- Network 最大化

而追求：

> 长期保持高质量行动能力。

它允许冒险。

允许努力。

允许承担压力。

允许冲刺。

允许失败。

允许探索。

甚至允许为了关键晋升暂时透支一些状态。

但是：

> 不能让任何一次短期成功摧毁继续发展的能力。

最终理想循环是：

**保持健康**

→

**积累有效 Network**

→

**持续提升 Skill**

→

**提前准备绩效**

→

**主动抓住关键机会**

→

**完成晋升**

→

**恢复状态**

→

**继续下一轮长期积累**

整个 Skill 的核心不是“永远选择最安全的选项”，而是：

> **在确保不会崩盘的前提下，敢于主动争取高价值机会。**
